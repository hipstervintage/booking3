```python
toplam_saniye= int  (input("Lütfen toplam saniye sayısını giriniz: "))
saat = (toplam_saniye//3600)
kalan_saniye = (toplam_saniye%3600)
dakika = (kalan_saniye//60)
saniye = (kalan_saniye%60)

print(f"{toplam_saniye} saniye girdiniz bu şuna eşit: {saat} saat, {dakika} ve {saniye} değerine eşittir." )
```
