```python
fiyat= float  (input("Lütfen ürünün birim fiyatını giriniz: "))
indirim_yüzdesi = float (input("Lütfen ürünün indirim yüzdesini giriniz:"))
adet = int (input ("Lütfen ürün adetini giriniz:"))
birim_indirim_miktari = fiyat*indirim_yüzdesi/100
indirimli_birim_fiyat = fiyat-birim_indirim_miktari

toplam_tutar= adet*indirimli_birim_fiyat
toplam_kazanc_tasarruf= birim_indirim_miktari * adet
print(f"Toplam tutar:{toplam_tutar} TL ")
print(f"İndirimli birim fiyat : {indirimli_birim_fiyat: .2f} TL" )
print(f"Toplam Ödenecek Tutar : {toplam_tutar: .2f} TL")
print(f"İndirimden Kazanılan Toplam Tasarruf : {toplam_kazanc_tasarruf: .2f} TL")
```
