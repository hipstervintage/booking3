```python
sayi= float(input ("Lütfen negatif, bir sayı giriniz"))
if sayi > 0:
      print ("girdiğiniz sayı pozitiftir")
elif sayi < 0:
      print ("girdiğiniz sayı negatiftir")
else:
        print ( "Girdiğiniz sayı sıfırdır")
```
